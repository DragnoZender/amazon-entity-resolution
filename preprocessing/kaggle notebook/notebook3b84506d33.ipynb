```python
import re
import unicodedata

def normalize_text(text):
    """
    Basic normalization for business names and addresses.
    Does not translate, correct spelling, or add information.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # 1. Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # 2. Convert to lowercase
    text = text.lower()

    # 3. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # 4. Normalize common punctuation to spaces
    text = re.sub(r"[.,;:!?()\[\]{}]", " ", text)

    # 5. Normalize hyphens/slashes to spaces
    text = re.sub(r"[-/]", " ", text)

    # 6. Normalize whitespace again
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
LEGAL_SUFFIX_PATTERNS = {
    r"\bprivate limited\b": "pvt ltd",
    r"\bprivate ltd\b": "pvt ltd",
    r"\bpvt limited\b": "pvt ltd",
    r"\bpvt ltd\b": "pvt ltd",
    r"\bpvt\b": "pvt",
    r"\bincorporated\b": "inc",
    r"\bcorporation\b": "corp",
    r"\blimited liability company\b": "llc",
}

def normalize_business_name(text):
    """
    Normalize a business name using the basic text normalization
    plus canonicalization of common legal suffix forms.
    """

    text = normalize_text(text)

    if not text:
        return ""

    for pattern, replacement in LEGAL_SUFFIX_PATTERNS.items():
        text = re.sub(pattern, replacement, text)

    # Normalize whitespace again
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def normalize_business_address(text):
    """
    Normalize a business address.

    Preserves numbers and multilingual text.
    Does not geocode, translate, or add missing information.
    """

    text = normalize_text(text)

    if not text:
        return ""

    for pattern, replacement in ADDRESS_ABBREVIATIONS.items():
        text = re.sub(pattern, replacement, text)

    # Normalize whitespace again
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def preprocess_dataframe(df):
    """
    Add normalized business name and address columns
    while preserving all original columns.
    """

    df = df.copy()

    df["business_name_normalized"] = (
        df["business_name"]
        .map(normalize_business_name)
    )

    df["business_address_normalized"] = (
        df["business_address"]
        .map(normalize_business_address)
    )

    return df
```

```python
from difflib import SequenceMatcher

def similarity(a, b):
    return SequenceMatcher(
        None,
        str(a),
        str(b)
    ).ratio()
```

```python
# Select ground-truth rows that have at least one match
gt_sample = ground_truth[
    ground_truth["matched_entity_ids"].notna() &
    (ground_truth["matched_entity_ids"].str.strip() != "")
].sample(
    100,
    random_state=42
)

results = []

for _, row in gt_sample.iterrows():

    s1_id = row["source1_entity_id"]

    matched_ids = [
        x.strip()
        for x in row["matched_entity_ids"].split(",")
        if x.strip()
    ]

    s1 = train_s1[
        train_s1["entity_id"] == s1_id
    ]

    if s1.empty:
        continue

    s1_row = s1.iloc[0]

    for match_id in matched_ids:

        # Match can belong to Source 2 or Source 3
        match = train_s2[
            train_s2["entity_id"] == match_id
        ]

        source = "S2"

        if match.empty:
            match = train_s3[
                train_s3["entity_id"] == match_id
            ]
            source = "S3"

        if match.empty:
            continue

        match_row = match.iloc[0]

        results.append({
            "source1_id": s1_id,
            "matched_id": match_id,
            "source": source,

            "name_raw_similarity": similarity(
                s1_row["business_name"],
                match_row["business_name"]
            ),

            "name_normalized_similarity": similarity(
                s1_row["business_name_normalized"],
                match_row["business_name_normalized"]
            ),

            "address_raw_similarity": similarity(
                s1_row["business_address"],
                match_row["business_address"]
            ),

            "address_normalized_similarity": similarity(
                s1_row["business_address_normalized"],
                match_row["business_address_normalized"]
            )
        })

similarity_results = pd.DataFrame(results)

print("Number of positive pairs:", len(similarity_results))

similarity_results.head(10)
```

```python
print("Average similarity improvement")
print("=" * 50)

print(
    "Name:",
    similarity_results["name_normalized_similarity"].mean()
    - similarity_results["name_raw_similarity"].mean()
)

print(
    "Address:",
    similarity_results["address_normalized_similarity"].mean()
    - similarity_results["address_raw_similarity"].mean()
)

print("\nAverage similarities")
print("=" * 50)

print(
    similarity_results[
        [
            "name_raw_similarity",
            "name_normalized_similarity",
            "address_raw_similarity",
            "address_normalized_similarity"
        ]
    ].mean()
)
```

```python
%%writefile preprocessing.py

import re
import unicodedata
import pandas as pd


# ============================================================
# Basic text normalization
# ============================================================

def normalize_text(text):
    """
    Basic normalization for business names and addresses.
    Does not translate, correct spelling, or add information.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Normalize punctuation
    text = re.sub(r"[.,;:!?()\[\]{}]", " ", text)

    # Normalize hyphens and slashes
    text = re.sub(r"[-/]", " ", text)

    # Normalize whitespace again
    text = re.sub(r"\s+", " ", text).strip()

    return text


# ============================================================
# Business name normalization
# ============================================================

LEGAL_SUFFIX_PATTERNS = {
    r"\bprivate limited\b": "pvt ltd",
    r"\bprivate ltd\b": "pvt ltd",
    r"\bpvt limited\b": "pvt ltd",
    r"\bpvt ltd\b": "pvt ltd",
    r"\bpvt\b": "pvt",
    r"\bincorporated\b": "inc",
    r"\bcorporation\b": "corp",
    r"\blimited liability company\b": "llc",
}


def normalize_business_name(text):
    """
    Normalize a business name using basic text normalization
    plus canonicalization of common legal suffix forms.
    """

    text = normalize_text(text)

    if not text:
        return ""

    for pattern, replacement in LEGAL_SUFFIX_PATTERNS.items():
        text = re.sub(pattern, replacement, text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


# ============================================================
# Address normalization
# ============================================================

ADDRESS_ABBREVIATIONS = {
    r"\broad\b": "rd",
    r"\bstreet\b": "st",
    r"\bavenue\b": "ave",
    r"\bboulevard\b": "blvd",
    r"\bdrive\b": "dr",
    r"\blane\b": "ln",
    r"\bcircle\b": "cir",
    r"\bhighway\b": "hwy",
    r"\bparkway\b": "pkwy",
    r"\bcourt\b": "ct",
    r"\bplace\b": "pl",
    r"\bterrace\b": "ter",
    r"\bbroadway\b": "bway",
}


def normalize_business_address(text):
    """
    Normalize a business address.

    Preserves numbers and multilingual text.
    Does not geocode, translate, or add missing information.
    """

    text = normalize_text(text)

    if not text:
        return ""

    for pattern, replacement in ADDRESS_ABBREVIATIONS.items():
        text = re.sub(pattern, replacement, text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


# ============================================================
# DataFrame preprocessing
# ============================================================

def preprocess_dataframe(df):
    """
    Add normalized business name and address columns
    while preserving the original columns.
    """

    df = df.copy()

    df["business_name_normalized"] = (
        df["business_name"].map(normalize_business_name)
    )

    df["business_address_normalized"] = (
        df["business_address"].map(normalize_business_address)
    )

    return df
```
